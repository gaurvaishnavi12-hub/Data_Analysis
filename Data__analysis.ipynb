### File import

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/Global-Superstore.csv", encoding='latin1')

print("Dataset loaded successfully!")
print("Shape of dataset:", df.shape)

df.head()
```

### Dataset Summary

```python
print("Column Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all").T)
```

### Data Cleaning


```python
# Check missing values
print("Missing values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# Check data types
print("\nData types:")
print(df.dtypes)
```

### Extra Row Delete

```python
# Create a working copy
df_clean = df.copy()

# Remove mostly empty extra columns
extra_columns = [
    "Unnamed: 24",
    "Unnamed: 25",
    "Sales_10%",
    "Unnamed: 27"
]

df_clean = df_clean.drop(
    columns=extra_columns,
    errors="ignore"
)

print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)

print("\nRemaining columns:")
print(df_clean.columns.tolist())
```

### Date Column Convert

```python
df_clean["Order Date"] = pd.to_datetime(
    df_clean["Order Date"],
    errors="coerce"
)

df_clean["Ship Date"] = pd.to_datetime(
    df_clean["Ship Date"],
    errors="coerce"
)

print(df_clean[["Order Date", "Ship Date"]].dtypes)

print("\nOrder Date range:")
print(df_clean["Order Date"].min())
print(df_clean["Order Date"].max())
```

### MIssing values and Data Quality

```python
print("Missing values after cleaning:")
print(df_clean.isnull().sum())

print("\nDuplicate rows after cleaning:")
print(df_clean.duplicated().sum())

print("\nInvalid dates:")
print("Order Date:", df_clean["Order Date"].isna().sum())
print("Ship Date:", df_clean["Ship Date"].isna().sum())
```

### Sales Analysis

```python
total_sales = df_clean["Sales"].sum()
total_profit = df_clean["Profit"].sum()
total_quantity = df_clean["Quantity"].sum()
average_discount = df_clean["Discount"].mean()
total_orders = df_clean["Order ID"].nunique()

print("GLOBAL SUPERSTORE SALES ANALYSIS")
print("-" * 40)

print(f"Total Sales: {total_sales:,.2f}")
print(f"Total Profit: {total_profit:,.2f}")
print(f"Total Quantity Sold: {total_quantity:,}")
print(f"Average Discount: {average_discount:.2%}")
print(f"Unique Orders: {total_orders:,}")
```

### Category Wise Sales and Profit analysis

```python
category_analysis = df_clean.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
).sort_values("Total_Sales", ascending=False)

print(category_analysis.round(2))
```

### Category-Wise Sales Chart

```python
category_analysis["Total_Sales"].plot(
    kind="bar",
    figsize=(9, 5),
    title="Category-wise Total Sales"
)

plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Category-Wise Profit Chart

```python
category_analysis["Total_Profit"].plot(
    kind="bar",
    figsize=(9, 5),
    title="Category-wise Total Profit"
)

plt.xlabel("Product Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Region-Wise Analysis

```python
region_analysis = df_clean.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum")
).sort_values("Total_Sales", ascending=False)

print(region_analysis.round(2))
```

### Region-Wise Chart

```python
region_analysis["Total_Sales"].sort_values().plot(
    kind="barh",
    figsize=(10, 7),
    title="Region-wise Total Sales"
)

plt.xlabel("Total Sales")
plt.ylabel("Region")
plt.tight_layout()
plt.show()
```

### Region-Wise Profit

```python
region_analysis["Total_Profit"].sort_values().plot(
    kind="barh",
    figsize=(10, 7),
    title="Region-wise Total Profit"
)

plt.xlabel("Total Profit")
plt.ylabel("Region")
plt.tight_layout()
plt.show()
```

### Montly Sales Trend

```python
monthly_sales = (
    df_clean
    .set_index("Order Date")["Sales"]
    .resample("MS")
    .sum()
)

print(monthly_sales.head(12).round(2))
```

### Montly Sales Line Chart

```python
plt.figure(figsize=(14, 5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend (2011–2014)")
plt.xlabel("Order Date")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Top 10 Product Analysis

```python
top_products = (
    df_clean.groupby("Product Name")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
    .sort_values("Total_Sales", ascending=False)
    .head(10)
)

print(top_products.round(2))
```

### Top 10 Product Chart

```python
top_products["Total_Sales"].sort_values().plot(
    kind="barh",
    figsize=(12, 7),
    title="Top 10 Products by Sales"
)

plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()
```

### DIscount And Profit Analysis

```python
discount_analysis = df_clean.groupby("Discount").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Order_Rows=("Order ID", "count")
).sort_index()

print(discount_analysis.round(2))
```

### Scatter Plot

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df_clean,
    x="Discount",
    y="Profit",
    alpha=0.4
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

### Finding Loss-Manking Product

```python
product_profit = (
    df_clean.groupby("Product Name")["Profit"]
    .sum()
    .sort_values()
)

loss_products = product_profit[product_profit < 0]

print("Number of loss-making products:", len(loss_products))
print("\nBottom 10 Products by Profit:")
print(loss_products.head(10).round(2))
```

### Customer segment Analysis

```python
segment_analysis = df_clean.groupby("Segment").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Unique_Customers=("Customer ID", "nunique")
).sort_values("Total_Sales", ascending=False)

print(segment_analysis.round(2))
```

#### Sales Chart

```python
segment_analysis["Total_Sales"].plot(
    kind="bar",
    figsize=(8, 5),
    title="Sales by Customer Segment"
)

plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import os

os.makedirs("outputs/charts", exist_ok=True)
os.makedirs("outputs/tables", exist_ok=True)

# Save analysis tables
category_analysis.to_csv(
    "outputs/tables/category_analysis.csv"
)

region_analysis.to_csv(
    "outputs/tables/region_analysis.csv"
)

segment_analysis.to_csv(
    "outputs/tables/segment_analysis.csv"
)

top_products.to_csv(
    "outputs/tables/top_10_products.csv"
)

monthly_sales.to_csv(
    "outputs/tables/monthly_sales.csv"
)

# Save cleaned dataset
df_clean.to_csv(
    "outputs/cleaned_global_superstore.csv",
    index=False
)

print("Analysis tables and cleaned dataset saved!")
```
